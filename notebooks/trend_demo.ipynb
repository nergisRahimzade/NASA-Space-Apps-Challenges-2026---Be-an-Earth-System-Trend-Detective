# ESD-D03-AI: Mann-Kendall Trend Test Baseline Analysis
This notebook demonstrates how to load our simulated climate time-series data, execute the non-parametric Mann-Kendall trend test using `pymannkendall`, and explicitly define the statistical meaning of each output parameter.

```python
import os
import pandas as pd
import pymannkendall as mk

# 1. Load the generated fake time-series data
data_path = os.path.join("..", "data", "fake_series.csv")
df = pd.read_csv(data_path)
print(f"Loaded {len(df)} months of climate simulation data.")
```

```python
# 2. Execute the standard Mann-Kendall Trend Test
result = mk.original_test(df["raw_value"])

# Print raw object outputs
print("--- Raw Mann-Kendall Test Results ---")
print(f"trend: {result.trend}")
print(f"p:     {result.p:.6f}")
print(f"tau:   {result.tau:.6f}")
print(f"slope: {result.slope:.6f}")
```

## Output Definitions (One-Line Technical Documentation)

* **trend (increasing / decreasing / no trend):** Specifies the directional movement of the time-series based on whether the total count of positive step changes significantly outnumbers negative ones.
* **p (p-value):** Measures the probability that the observed trend occurred by random chance, where a value below 0.05 rejects the null hypothesis of no trend.
* **tau (Kendall Tau coefficient):** Represents a rank correlation ranging from -1 to +1, indicating the overall monotonic strength and direction of the relationship over time.
* **slope (Sen's Slope):** Quantifies the median rate of change per time step across all paired data points, representing the physical velocity of the trend (e.g., change units per month).
